```python
import numpy as np
import matplotlib.pyplot as plt
import json
%matplotlib widget
```

```python
 with open("solve_3d_pts.json", "r") as f:
    my_data = json.load(f)
```

```python
nrows = 1
ncols = 1
fig, axs = plt.subplots(nrows, ncols)

pts_background = np.array(my_data["background_pts"]).transpose()
pts_crvs = []
for crv in my_data["crv_pts"]:
    pts_crvs.append(np.array(crv).transpose())
    print(f"{pts_crvs[-1].shape}")

cols = ["r-", "g:", "b--"]
axs.scatter(pts_background[0,:], pts_background[1,:], pts_background[2,:], c='k', marker='x')
for crv, col in zip(pts_crvs, cols):
    axs.plot(crv[0, :], crv[1, :], crv[2, :], col)

axs.set_aspect('equal')
```
